## Baseline Metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

actual = baseline_df['units_sold']
predicted = baseline_df['baseline_prediction']

mae = mean_absolute_error(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))

mask = actual != 0
mape = np.mean(
    np.abs((actual[mask] - predicted[mask]) / actual[mask])
) * 100

print("Baseline MAE:", mae)
print("Baseline RMSE:", rmse)
print("Baseline MAPE:", mape, "%")
```

## Insight

## The baseline model shows relatively high prediction error, indicating that a more advanced machine learning model may improve demand forecasting accuracy.

## Machine Learning Forecasting

## Features and Targets

```python
baseline_df.dtypes
```

```python
features = [
    'unit_price',
    'promo_flag',
    'unit_cost',
    'list_price',
    'week',
    'month',
    'is_holiday',
    'on_order_units',
    'lead_time_days',
    'year',
    'day_of_week',
    'lag_1',
    'lag_7',
    'rolling_7',
    'inventory_ratio',
    'reorder_gap',
    'on_hand_units',
    'reorder_point',
    'stockout_risk_flag',
    'overstock_risk_flag'
]

X = baseline_df[features]
y = baseline_df['units_sold']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

## Time-Based Train/Test Split

```python

baseline_df = baseline_df.sort_values('date').reset_index(drop=True)
split_index = int(len(baseline_df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## Train the Random Forest Model

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
```

```python
print("Random Forest model trained successfully.")
```

## Machine Learning Prediction

```python
y_pred = rf_model.predict(X_test)

print("Predictions generated successfully.")
print("Number of predictions:", len(y_pred))
```

## Actual vs ML Predicted Demand

```python
comparison_df = pd.DataFrame({
    'Actual_Demand': y_test.values,
    'ML_Predicted_Demand': y_pred
})

comparison_df.head(10)
```

## Machine Learning Model Performance

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

ml_mae = mean_absolute_error(y_test, y_pred)
ml_rmse = np.sqrt(mean_squared_error(y_test, y_pred))

mask = y_test != 0
ml_mape = np.mean(
    np.abs((y_test[mask] - y_pred[mask]) / y_test[mask])
) * 100

print("ML Model MAE:", ml_mae)
print("ML Model RMSE:", ml_rmse)
print("ML Model MAPE:", ml_mape, "%")
```

## Insight

## Risk Scoring

```python
baseline_df["inventory_gap"]=baseline_df["on_hand_units"]-baseline_df["final_prediction"]
```

```python
baseline_df[[
    "on_hand_units",
    "final_prediction",
    "inventory_gap"
]].head()
```

```python
def classify_risk(row):
    if row["inventory_gap"] < 0:
        return "High Stockout Risk"
    elif row["on_hand_units"] >= 2 * row["final_prediction"]:
        return "Overstock Risk"
    else:
        return "Healthy Inventory"

baseline_df["risk_category"] = baseline_df.apply(classify_risk, axis=1)
```

```python
risk_score_map = {
    "Healthy Inventory": 0,
    "Overstock Risk": 1,
    "High Stockout Risk": 2
}

baseline_df["risk_score"] = baseline_df["risk_category"].map(risk_score_map)
```

```python
action_map = {
    "Healthy Inventory": "Maintain Current Stock",
    "Overstock Risk": "Reduce Inventory / Increase Promotion",
    "High Stockout Risk": "Reorder Immediately"
}

baseline_df["recommended_action"] = baseline_df["risk_category"].map(action_map)
```
